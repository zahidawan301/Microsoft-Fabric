```python
entities = ["customers","orders","sales","products","churn"]
```

```python
dfs = {}

for i in entities:
    dfs[i] = spark.read.csv(f"Files/Bronze/{i}.csv",header=True,inferSchema=True)

    print(f"{i} dataframe is created")
```

```python
display(dfs["customers"].limit(5))
```

```python
dfs.keys()
```

```python
for entity, df in dfs.items():
    df.write.format("delta").mode("overwrite").saveAsTable(f"Silver.{entity}")
    print(f"Delta table created: Silver.{entity}")
```
